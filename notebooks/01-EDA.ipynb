This notebook analyzes grain bin environmental factors (grain temperature, moisture content, ambient NOAA weather) to establish safe storage boundaries under ASAE Standards D245.5.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset generated by AgroShield pipeline
df = pd.read_csv('../data/samples/iowa_silo_telemetry_sample.csv')
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Key statistical distribution
df.describe()
```

```python
# Grain moisture vs Risk Level analysis
plt.figure(figsize=(10, 5))
sns.boxplot(x='risk_level', y='grain_moisture_pct', data=df, palette='Set2')
plt.title('Grain Moisture Distribution Across Risk Categories (0: Safe, 1: Warning, 2: Critical)')
plt.xlabel('Risk Level Category')
plt.ylabel('Grain Moisture (%)')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Thermal Gradient vs EMC Correlation Heatmap
plt.figure(figsize=(8, 6))
numeric_cols = ['grain_temp_c', 'grain_moisture_pct', 'ambient_temp_c', 'ambient_rh_pct', 'emc_pct', 'temp_gradient_c', 'risk_level']
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Feature Correlation Matrix — Silo Spoilage Determinants')
plt.show()
```
